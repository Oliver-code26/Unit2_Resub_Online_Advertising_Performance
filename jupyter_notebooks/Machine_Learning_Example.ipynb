I will ask Copilot using the comments (#) to ask for chances of having a four (4:x) out of my hand (7 dominoes). What is the common number (Mode) in the set, possibly counting the dots? 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import sys
import sklearn
```

```python

total_dots = 0
total_dominoes = 0

for a in range(7):
    for b in range(a, 7):
        dots = a + b
        total_dots += dots
        total_dominoes += 1

print("Total dominoes:", total_dominoes)
print("Total dots:", total_dots)
```

```python
#show me a set of dominoes
dominoes = []
for a in range(7):
    for b in range(a, 7):
        dominoes.append((a, b))
print("Dominoes:", dominoes)
```

Brilliant! This is the full set of dominoes.

```python
#how many dominoes have 5 dots?
count_5_dots = 0
for a in range(7):
    for b in range(a, 7):
        dots = a + b
        if dots == 5:
            count_5_dots += 1
print("Dominoes with 5 dots:", count_5_dots)
```

Interesting, I was more asking how many dominoes has a 5 on it but the machine (?) has counted how many dominoes add up to 5! Which are 4:1, 5:0 & 3:2.

```python
#show me how many dominoes have a 5 in it
dominoes_with_5 = [domino for domino in dominoes if 5 in domino]
count_5 = len(dominoes_with_5)
print("Number of dominoes with a 5:", count_5)
```

That's more like it. All that needed changing was how I asked Copilot, as I thought previously and we can see that yes, there are 7 dominoes with a 5 in it.

5:5, 5:4, 5:3, 5:2, 5:1 & 5:0

## Mode, Median & Mean 

```python
#show me the mode of dominoes
from collections import Counter

domino_counts = Counter(domino for domino in dominoes)
mode_domino = domino_counts.most_common(1)[0][0]
print("Mode domino:", mode_domino)
```

```python
#show me the median of dominoes
domino_values = [a + b for a, b in dominoes]
domino_values.sort()
n = len(domino_values)
if n % 2 == 0:
    median = (domino_values[n//2 - 1] + domino_values[n//2]) / 2
else:
    median = domino_values[n//2]
print("Median domino value:", median)
```

```python
#show me the mean of dominoes
mean = total_dots / total_dominoes
print("Mean domino value:", mean)
```

## Probability Test

Now that I have asked for Mode, Median, Mean and got How many dominoes add up to five. I will attempt some probabilities.

```python
#show me the probability of drawing dominoes with a 2 out of a hand of 7
dominoes_with_2 = [domino for domino in dominoes if 2 in domino]
probability = len(dominoes_with_2) / len(dominoes)
print("Probability of drawing a domino with a 2:", probability)
```

Fantastic! 25% chance of drawing dominoes with a 2:x on it! Obviously this is not a guarenteed as I've only got 7/28 to make a hand, my opponent would also have 7/28 and leaving 14/28 idling (Left in the Boneyard)

28 = 100% (1)

21 = 75% (0.75)

14 = 50% (0.50)

7 = 25% (0.25)


```python
#show me the probability of drawing dominoes with a 3 out of a hand of 7 and another hand of 7
from itertools import combinations

hand1 = [domino for domino in dominoes if 3 in domino]
hand2 = [domino for domino in dominoes if 3 in domino]

probabilities = []
for domino1 in hand1:
    for domino2 in hand2:
        if domino1 != domino2:
            probabilities.append(1 / (len(dominoes) * (len(dominoes) - 1)))

final_probability = sum(probabilities)
print("Probability of drawing a domino with a 3 from two hands of 7:", final_probability)
```

```python
#show me the probability of two hands drawing '6, 6' from a set of dominoes
dominoes_with_6_6 = [domino for domino in dominoes if domino == (6, 6)]
probability_6_6 = len(dominoes_with_6_6) / len  (dominoes)
print("Probability of drawing '6, 6' from two hands of 7:", probability_6_6)
```
